```python
import pest

dataset = pest.PynbodyDataset("tests/data/illustris_tng/TNG50-1/snapdir_099/snap_099")

len(dataset)
```

```python
subhalo = dataset[0]
subhalo.keys()
```

```python
render = pest.RenderStars()
render.bind(dataset)
result = render(subhalo)
image = result["image"]
```

```python
import matplotlib.pyplot as plt

plt.rcParams["text.usetex"] = False  # pynbody enables LaTeX labels, which needs a latex install
plt.imshow(image)
plt.axis("off")
plt.title(f"Subhalo {subhalo['subhalo_id']}")
plt.show()
```
