```python
from pest import IllustrisDownloader, Selector, PropertyType

mass_min = 10**11.99 / 1e10 * 0.704
mass_max = 10**12.01 / 1e10 * 0.704

illustris_downloader = IllustrisDownloader()
subhalos = illustris_downloader.get_subhalos(
    selector=Selector(property=PropertyType.MASS, min_value=mass_min, max_value=mass_max)
)
subhalos["count"]
```

```python
ids = [subhalos["results"][i]["id"] for i in range(5)]
ids
```

```python
import illustris_python as il

basePath = "/home/doserbd/data/Illustris-3/output"
fields = ["SubhaloMass", "SubhaloSFRinRad"]
subhalos = il.groupcat.loadSubhalos(basePath, 135, fields=fields)
```
