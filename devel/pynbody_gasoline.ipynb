```python
import pynbody

pynbody.config["number_of_threads"] = 8

snap_path = "testdata/gasoline_ahf/g15784.lr.01024.gz"
f = pynbody.load(snap_path)
f.physical_units()
```

```python
h = f.halos()
main_halo = h[0]
```

```python
pynbody.analysis.center(main_halo)
pynbody.plot.image(main_halo.gas, width=100, cmap="Blues")
```

```python
pynbody.plot.stars.render(main_halo.stars, width="50 kpc", with_dust=True, mag_range=[18, 26])
```
