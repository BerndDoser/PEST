```python
import pynbody

snap_path = "/urz/gpuscratch/its/data/Illustris/TNG50-1/snapdir_099/snap_099"
f = pynbody.load(snap_path)
f.physical_units()
print(f.loadable_keys())
print(f.gas.loadable_keys())
```

```python
h = f.halos(subhalos=True)
```

```python
print("Number of halos:", len(h))
print("Number of particles in the first halo:", len(h[0]))
print(h[0].properties.keys())
```

```python
subhalo = h[0].subhalos[0]
print("Number of subhalos in the first halo:", len(subhalo))
print("Number of particles in the first subhalo of the first halo:", len(subhalo[0]))
print(subhalo[0].properties)
```
